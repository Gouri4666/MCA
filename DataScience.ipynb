NLTK

```python
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
text = "Hello World"
print(word_tokenize(text))
```

Beautiful Soup

Beautiful Soup (bs4) is a Python library used for web scraping. It helps extract data from HTML and XML pages.

1. Parsing and extrcating data

```python
from bs4 import BeautifulSoup
html_doc = """
<html>
<head>
  <title>Page Title</title>
</head>
<body>
  <h1>Welcome to BeautifulSoup4</h1>
  <p>This is a sample page.</p>
  <ul>
    <li>Item 1</li>
    <li>Item 2</li>
    <li>Item 3</li>
  </ul>
</body>
</html>
"""
soup = BeautifulSoup(html_doc, 'html.parser')
header = soup.find('h1')
print(header.text)
items = soup.find_all('li')
for item in items:
    print(item.text)
```

2. Format HTML

```python
from bs4 import BeautifulSoup

html = "<html><body><h1>Hello</h1></body></html>"

soup = BeautifulSoup(html, "html.parser")

print(soup.prettify())
```

1. DNA sequeance operation

```python
from Bio.Seq import Seq

dna = Seq("ATGGCC")

print("DNA:", dna)
print("Length:", len(dna))
print("RNA:", dna.transcribe())
print("Protein:", dna.translate())
print("Complement:", dna.complement())
print("Reverse Complement:", dna.reverse_complement())
print("Count of G:", dna.count("G"))
```

2. DNA to RNA

```python
from Bio.Seq import Seq

dna = Seq("ATGGCC")

print("DNA:", dna)
print("RNA:", dna.transcribe())
print("Protein:", dna.translate())
```

3. Sequence comparison and slicing

```python
from Bio.Seq import Seq

dna1 = Seq("ATGCGA")
dna2 = Seq("ATGCAA")

print("DNA1:", dna1)
print("DNA2:", dna2)
print("Are Equal?", dna1 == dna2)
print("First 3 Bases:", dna1[:3])
print("Last 3 Bases:", dna1[-3:])
```
